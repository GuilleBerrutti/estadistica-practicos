```r
#Ejercicio 2

# Opción 1: Cálculo analítico
letras <- c("S", "A", "B", "A", "N", "A")
n_total <- length(letras)
frecuencias <- table(letras)

resultado <- factorial(n_total) / prod(factorial(frecuencias))
print(resultado) # 120

# Opción 2: Generando las permutaciones únicas
library(gtools)
matriz <- permutations(n = 6, r = 6, v = letras, set = FALSE)
palabras <- apply(matriz, 1, paste, collapse = "")
palabras_unicas <- unique(palabras)
length(palabras_unicas) # 120
```

```r
#Ejercicio 4

# Opción 1: Cálculo analítico
total_numeros <- 9 * 10^5
solo_impares <- 5^6

resultado <- total_numeros - solo_impares
print(resultado) # 884375

# Opción 2: Verificación por fuerza bruta
# Generamos todos los números de 6 cifras (de 100000 a 999999)
numeros <- 100000:999999

# Función para verificar si un número tiene al menos una cifra par
tiene_par <- function(n) {
  digitos <- as.numeric(unlist(strsplit(as.character(n), "")))
  any(digitos %% 2 == 0)
}

# Contamos cuántos cumplen la condición
sum(sapply(numeros, tiene_par)) # 884375
```
