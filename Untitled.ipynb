```r
# 1. Función de cuantía
x_vals <- 0:2
f_X <- dhyper(x = x_vals, m = 3, k = 5, n = 2) # m=defectuosos, k=no defectuosos, n=muestra
names(f_X) <- x_vals
print(f_X)

# 2. P(X = 2)
p_todos_def <- dhyper(2, m = 3, k = 5, n = 2)
print(p_todos_def)

# 3. Función de distribución y gráfico
F_X <- stepfun(x_vals, c(0, phyper(x_vals, m = 3, k = 5, n = 2)))
plot(F_X, main = "Función de Distribución F_X", xlab = "x", ylab = "F(x)")
```
