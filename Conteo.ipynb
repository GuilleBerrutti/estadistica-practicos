```r
# Ejercicio 2

# Parte a) Total de listas posibles (Arreglo de 10 tomados de a 4)
total_listas <- prod(10:7) # o factorial(10) / factorial(6)
print(total_listas) # 5040

# Parte b.i) Un médico en la presidencia
# 3 opciones para Presidente * Arreglo de las 9 personas restantes para 3 puestos
b_i <- 3 * prod(9:7)
print(b_i) # 1512

# Parte b.ii) Exactamente un médico en la lista
# 4 cargos * 3 médicos * Arreglo de los 7 no-médicos para los 3 puestos restantes
b_ii <- 4 * 3 * prod(7:5)
print(b_ii) # 2520

# Parte b.iii) Al menos un médico en la lista
# Total de listas - Listas sin ningún médico (usando solo a los 7 no-médicos)
listas_sin_medicos <- prod(7:4)
b_iii <- total_listas - listas_sin_medicos
print(b_iii) # 4200
```

```r
# Ejercicio 9

# Datos iniciales:
# Vocales: U, I, E, A, O (5)
# Consonantes: M, R, C, L, G (5)

# a) Elegir 1 consonante Y 1 vocal (combinaciones simples de 5 tomadas de a 1)
a <- choose(5, 1) * choose(5, 1)
print(a) # 25

# b) Palabras de 2 letras formadas por 1 consonante y 1 vocal (25 pares * 2 ordenamientos)
b <- a * factorial(2)
print(b) # 50

# c) Palabras de 5 letras posibles (Arreglo sin repetición de 10 tomadas de a 5)
c <- prod(10:6) # o factorial(10) / factorial(5)
print(c) # 30,240

# d) Comienzan con M (M fija en el 1er lugar, elegimos y ordenamos 4 letras entre las 9 restantes)
d <- prod(9:6)
print(d) # 3,024

# e) Contienen M (M puede ir en cualquiera de las 5 posiciones * arreglo de las 9 restantes)
e <- 5 * prod(9:6)
print(e) # 15,120

# f) Comienzan con M y terminan con O (M y O fijas en extremos, 3 lugares para las 8 restantes)
f <- prod(8:6)
print(f) # 336

# g) Aparecen juntas M y O (Bloque MO u OM = 2 ordenes * 4 bloques posibles * arreglo de 8 tomadas de a 3)
g <- 2 * 4 * prod(8:6)
print(g) # 2,688

# h) Aparecen juntas M, O y U (Bloque de 3 letras = 3! ordenes * 3 bloques posibles * arreglo de 7 tomadas de a 2)
h <- factorial(3) * 3 * prod(7:6)
print(h) # 756

# i) Aparece el vocablo exacto MOU (Bloque fijo MOU = 1 orden * 3 posiciones * arreglo de 7 tomadas de a 2)
i_res <- 1 * 3 * prod(7:6)
print(i_res) # 126
```

```r
# Ejercicio 11

# Parte a)
triangulares <- 4
rectangulares_opciones <- 9^0 + 9^1 + 9^2 + 9^3

senales_a <- triangulares * rectangulares_opciones
print(senales_a) # 3280

# Parte b)
# Opción 1: Perder 3 triangulares (queda 1)
senales_perder_triangulares <- 1 * rectangulares_opciones
print(senales_perder_triangulares) # 820

# Opción 2: Perder 2 juegos rectangulares (pasa a ser sin repetición)
rect_sin_rep <- 1 + 9 + prod(9:8) + prod(9:7)
senales_perder_juegos <- 4 * rect_sin_rep
print(senales_perder_juegos) # 2344
```

```r
# Ejercicio 12

# Parte a) 9 tomados de a 5
a <- prod(9:5)
print(a) # 15120

# Parte b) 1er lugar fijo (el 2)
b <- prod(8:5)
print(b) # 1680

# Parte c) 1er lugar (2) y 3er lugar (5) fijos
c <- prod(7:5)
print(c) # 210

# Parte d) 1er dígito entre 1 y 8 (8 opciones)
d <- 8 * prod(8:5)
print(d) # 13440

# Parte e) Menores que 89763
e1 <- 7 * prod(8:5)     # Comienzan con 1..7 (11760)
e2 <- 7 * prod(7:5)     # Comienzan con 8 y 2do dígito 1..7 (1470)
e3 <- 6 * prod(6:5)     # Comienzan con 89 y 3er dígito 1..6 (180)
e4 <- 5 * 5             # Comienzan con 897 y 4to dígito 1..5 (25)
e5 <- 2                 # Comienzan con 8976 y 5to dígito 1..2 (2)

e_total <- e1 + e2 + e3 + e4 + e5
print(e_total) # 13437
```

```r
# Ejercicio 13

# Parte a)
# Al ser números primos (y el 1), cada combinación produce un producto único.
a <- choose(8, 5)
print(a) # 56

# Parte b)
conjunto_b <- c(1, 2, 3, 4, 6, 9, 11)

# Generamos todas las combinaciones de 5 elementos (cada columna es una combinación)
comb_b <- combn(conjunto_b, 5)

# Calculamos el producto de los 5 elementos de cada columna
productos_b <- apply(comb_b, 2, prod)

# Obtenemos los productos únicos
productos_unicos <- unique(productos_b)

# Contamos cuántos productos distintos hay
b <- length(productos_unicos)
print(b) # 19
```

```r
#Ejercicio 15

library(gtools)

# Generar todas las permutaciones de las cifras 1 a 6
perms <- permutations(n = 6, r = 6, v = 1:6)

# Convertir cada fila en un número de 6 dígitos
numeros <- apply(perms, 1, function(row) sum(row * 10^(5:0)))

# Calcular la suma de todos los números
suma_total <- sum(numeros)
print(suma_total) # 279999720
                 
```
